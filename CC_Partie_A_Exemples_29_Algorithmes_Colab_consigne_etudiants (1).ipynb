## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La **Régression Linéaire** (*Linear Regression*) est un algorithme de la famille de l'**apprentissage supervisé (régression)**. Son objectif est de modéliser la relation entre une variable cible continue $y$ et une ou plusieurs variables explicatives $X$ en ajustant une fonction linéaire.

*   **Intuition et étapes :** Données en entrée ($X$, $y$) $\rightarrow$ Apprentissage (calcul des coefficients qui minimisent la somme des carrés des résidus) $\rightarrow$ Prédiction d'une valeur continue en sortie $\hat{y}$ pour une nouvelle entrée.
*   **Formule mathématique essentielle :**
    $$\hat{y} = \beta_0 + \sum_{j=1}^{p} \beta_j X_j$$
    *   $\hat{y}$ : valeur prédite (variable dépendante).
    *   $X_j$ : variables explicatives (caractéristiques).
    *   $\beta_0$ : ordonnée à l'origine (biais, paramètre appris).
    *   $\beta_j$ : coefficients de régression (poids, paramètres appris).
*   **Avantages :** Très simple à interpréter et rapide à entraîner.
*   **Limites :** Sensible aux valeurs aberrantes (*outliers*) et suppose une relation strictement linéaire.

### **Schéma explicatif**
Voici une visualisation de la droite de régression ajustée sur la première dimension de nos données pédagogiques.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La régression **Ridge** (*Ridge Regression*) est une variante de la régression linéaire qui intègre une régularisation de type $L_2$. Elle appartient à la famille de l'**apprentissage supervisé (régression)**.

*   **Intuition et étapes :** Lorsque les variables d'entrée sont fortement corrélées, les coefficients de la régression classique peuvent devenir instables et sur-apprendre. Ridge pénalise la norme au carré des coefficients pour les maintenir petits.
*   **Formule mathématique essentielle (Fonction de coût) :**
    $$J(\beta) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \beta_j^2$$
    *   $\alpha$ : hyperparamètre de régularisation (force de la pénalité).
    *   $\beta_j$ : coefficients associés aux variables explicatives.
*   **Avantages :** Réduit efficacement le surapprentissage (*overfitting*) et gère la colinéarité.
*   **Limites :** Ne réduit pas les coefficients à zéro (ne fait pas de sélection de variables).

### **Schéma explicatif**
Visualisation de l'effet du paramètre de régularisation $\alpha$ sur la taille des coefficients.

```python
from sklearn.linear_model import Ridge
import matplotlib.pyplot as plt

alphas = [0.1, 10, 100, 1000]
coefs = []
for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(Xr_ent, yr_ent)
    coefs.append(ridge.coef_)

plt.figure(figsize=(8, 4))
plt.plot(alphas, coefs, marker='o')
plt.xscale('log')
plt.xlabel('Alpha (Force de pénalisation L2)')
plt.ylabel('Coefficients du modèle')
plt.title('Effet de la régularisation Ridge sur l\'amplitude des poids')
plt.grid(True)
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Finance :* Modélisation de la prime de risque d'un portefeuille d'actifs financiers où de nombreux facteurs macroéconomiques sont corrélés.
    2.  *Logistique :* Prédiction des temps de livraison avec de nombreuses variables géographiques et climatiques potentiellement redondantes.
*   **Nature des données :** Données numériques continues. Nécessite une normalisation préalable des variables car la pénalité est sensible à l'échelle.
*   **Cas peu approprié :** Cas où l'on souhaite un modèle très parcimonieux avec peu de variables actives (car Ridge garde toutes les variables).
*   **Métriques adaptées :** MAE, RMSE, $R^2$.

### **Analyse de l'expérience Python**
L'expérience utilise un pipeline combinant un centrage-réduction des données (`StandardScaler`) et une régression `Ridge` avec $\alpha=1.0$.
*   **Entrées :** `Xr_ent` et `yr_ent` standardisés.
*   **Hyperparamètres :** $\alpha = 1.0$ (contrôle la régularisation).
*   **Sortie :** Une MAE d'environ `9.37`.
*   **Conclusion :** La performance est très proche de celle de la régression linéaire classique, ce qui suggère que le niveau de colinéarité ou le bruit n'exigeait pas une forte pénalisation sur ce jeu de données.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La régression **Lasso** (*Least Absolute Shrinkage and Selection Operator*) est un modèle d'**apprentissage supervisé (régression)** qui applique une régularisation de type $L_1$.

*   **Intuition et étapes :** Contrairement à Ridge, Lasso force certains coefficients à s'annuler exactement lorsque la pénalité augmente. Cela permet de réaliser une sélection de caractéristiques intégrée.
*   **Formule mathématique essentielle :**
    $$J(\beta) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} |\beta_j|$$
    *   $\alpha$ : hyperparamètre de contrôle de la force de régularisation ($L_1$).
    *   $\beta_j$ : coefficients du modèle.
*   **Avantages :** Génère des modèles parcimonieux (sélection naturelle de variables explicatives) et réduit l'erreur de généralisation.
*   **Limites :** Si $p > n$, Lasso sélectionne au maximum $n$ variables. Si des variables sont fortement corrélées, il tend à en sélectionner une arbitrairement.

### **Schéma explicatif**
Voici une visualisation montrant comment l'évolution de $\alpha$ force certains coefficients du Lasso à s'annuler exactement, contrairement à Ridge.

```python
from sklearn.linear_model import Lasso
import matplotlib.pyplot as plt

alphas = [0.01, 0.1, 1, 10, 100]
coefs = []
for a in alphas:
    lasso = Lasso(alpha=a, max_iter=10000)
    lasso.fit(Xr_ent, yr_ent)
    coefs.append(lasso.coef_)

plt.figure(figsize=(8, 4))
plt.plot(alphas, coefs, marker='x')
plt.xscale('log')
plt.xlabel('Alpha (Force de pénalisation L1)')
plt.ylabel('Coefficients du modèle')
plt.title('Effet du Lasso sur les coefficients (Sparsité)')
plt.grid(True)
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Génomique :* Identification des gènes pertinents parmi des dizaines de milliers de candidats pour prédire la prédisposition à une maladie.
    2.  *Marketing :* Sélection des canaux de communication clés parmi un grand ensemble d'indicateurs d'audience pour modéliser le taux de conversion.
*   **Nature des données :** Données numériques. Standardisation préalable obligatoire pour éviter de pénaliser injustement les variables à grande échelle.
*   **Cas peu approprié :** Cas où toutes les variables explicatives ont un effet petit mais réel et doivent être conservées.
*   **Métriques adaptées :** $R^2$, MAE, RMSE.

### **Analyse de l'expérience Python**
Le code utilise un pipeline standardisé combinant un StandardScaler et un Lasso avec $\alpha=0.1$.
*   **Entrées :** `Xr_ent` et `yr_ent` standardisés.
*   **Hyperparamètres :** `alpha=0.1`, `max_iter=10000`.
*   **Sortie :** Un $R^2$ d'environ `0.986`.
*   **Conclusion :** Un $R^2$ de 98.6% montre que le Lasso explique presque la totalité de la variance de la cible, validant l'efficacité du modèle sur ces données.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La **Régression Polynomiale** est une extension de la régression linéaire appartenant à l'**apprentissage supervisé (régression)**.

*   **Intuition et étapes :** Elle projette les variables explicatives initiales dans un espace de dimension supérieure en calculant des combinaisons polynomiales, puis ajuste une régression linéaire sur ces nouvelles caractéristiques.
*   **Formule mathématique essentielle (Cas d'un degré 2 univarié) :**
    $$\hat{y} = \beta_0 + \beta_1 X + \beta_2 X^2$$
    *   $X^2$ : variable transformée.
    *   $\beta_i$ : paramètres ajustés par moindres carrés.
*   **Avantages :** Permet de capturer des relations courbes et non-linéaires complexes sans changer d'algorithme de résolution.
*   **Limites :** Risque de surapprentissage important à mesure que le degré augmente, et instabilité aux bornes des intervalles d'estimation.

### **Schéma explicatif**
Comparaison visuelle d'un ajustement linéaire simple et d'un ajustement de degré 2.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Données simulées non-linéaires
x_curve = np.sort(generateur.uniform(-3, 3, 80)).reshape(-1, 1)
y_curve = 0.5 * (x_curve**2) + x_curve + generateur.normal(0, 1, (80, 1))

# Modèle Linéaire simple
reg_lin = LinearRegression().fit(x_curve, y_curve)

# Modèle Polynomial degré 2
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x_curve)
reg_poly = LinearRegression().fit(x_poly, y_curve)

plt.figure(figsize=(8, 4))
plt.scatter(x_curve, y_curve, color='gray', alpha=0.7, label='Données')
plt.plot(x_curve, reg_lin.predict(x_curve), color='red', label='Linéaire (Degré 1)')
plt.plot(x_curve, reg_poly.predict(x_poly), color='green', linewidth=2, label='Polynomial (Degré 2)')
plt.legend()
plt.title('Régression Polynomiale vs Régression Linéaire')
plt.show()
```

### **Contextes d'utilisation**

    1.  *Agronomie :* Modélisation du rendement d'une culture en fonction de l'apport en engrais (courbe en cloche d'optimum).
    2.  *Énergie :* Évolution de la consommation électrique d'un pays en fonction de la température extérieure (forte hausse par grand froid et par canicule).
*   **Nature des données :** Variables continues numériques à relations non-linéaires lisses.
*   **Cas peu approprié :** Données comportant des sauts brusques, des discontinuités ou des changements brusques de régime.
*   **Métriques adaptées :** MAE, RMSE, $R^2$.

### **Analyse de l'expérience Python**
L'expérience utilise un pipeline associant StandardScaler, PolynomialFeatures(degree=2) et une régulation Ridge avec $\alpha=10$.
*   **Entrées :** `Xr_ent` et `yr_ent`.

*   **Sortie :** Une MAE d'environ `10.90`.
*   **Conclusion :** La MAE est légèrement supérieure à celle de la régression linéaire simple (9.35). Cela indique que sur ce jeu de données purement linéaire, l'ajout de termes polynomiaux de degré 2 a complexifié inutilement le modèle, menant à un léger surapprentissage.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'**Arbre de Régression** (*Decision Tree Regressor*) est un algorithme d'**apprentissage supervisé (régression)** non-paramétrique.

*   **Intuition et étapes :** Il segmente l'espace des données d'entrée en régions homogènes (rectangles/hyperplans) à l'aide de coupures binaires successives, puis prédit pour chaque région la valeur moyenne des échantillons qui s'y trouvent.
*   **Formule mathématique essentielle (Critère de division) :**
    $$\text{MSE} = \frac{1}{N} \sum_{i \in R} (y_i - \bar{y}_R)^2$$
    L'arbre cherche le seuil de coupure qui minimise la somme des variances (ou MSE) des deux sous-régions créées.
*   **Avantages :** Facile à interpréter visuellement et ne requiert aucun prétraitement d'échelle (normalisation).
*   **Limites :** Très instable (forte variance) et sujet à un surapprentissage rapide si la profondeur maximale n'est pas limitée.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La **Forêt Aléatoire** pour la régression (*Random Forest Regressor*) est une méthode d'ensemble de la famille de l'**apprentissage supervisé (régression)**.

*   **Intuition et étapes :** Elle entraîne un grand nombre d'arbres de décision de manière indépendante en utilisant le mécanisme du *Bagging* (échantillonnage par bootstrap sur les lignes) et en sélectionnant un sous-ensemble aléatoire de caractéristiques (*features*) pour chaque division. La prédiction finale est la moyenne de toutes les prédictions individuelles.
*   **Formule mathématique essentielle :**
    $$\hat{y} = \frac{1}{B} \sum_{b=1}^{B} T_b(X)$$
    *   $B$ : nombre total d'arbres générés dans la forêt (hyperparamètre).
    *   $T_b(X)$ : prédiction de l'arbre individuel $b$.
*   **Avantages :** Très robuste face au surapprentissage (grâce à la diversité des arbres) et ne nécessite que peu de réglage d'hyperparamètres.
*   **Limites :** Le modèle final est une boîte noire difficile à interpréter et peut être lent en temps d'exécution lors de la prédiction si l'ensemble contient trop d'arbres.

### **Schéma explicatif**

*   **Cas peu approprié :** Prédiction de données temporelles à forte tendance (l'arbre ne peut pas extrapoler de valeurs supérieures à celles vues à l'entraînement).

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
Le **Gradient Boosting** pour la régression (*Gradient Boosting Regressor*) est une technique d'ensemble de la famille de l'**apprentissage supervisé (régression)**.

*   **Intuition et étapes :** Contrairement au Bagging (Forêt Aléatoire), il construit des arbres séquentiellement. Chaque nouvel arbre est entraîné pour corriger les erreurs (les résidus) du modèle combiné précédent en suivant la direction négative du gradient de la fonction de perte.
*   **Formule mathématique essentielle :**
    $$F_m(X) = F_{m-1}(X) + \eta \cdot h_m(X)$$
    *   $F_m(X)$ : modèle combiné à l'étape $m$.
    *   $h_m(X)$ : nouvel estimateur faible (arbre de régression ajusté sur les résidus).
    *   $\eta$ : taux d'apprentissage (*learning rate*, hyperparamètre).
*   **Avantages :** Précision prédictive souvent supérieure à la majorité des algorithmes concurrents si bien ajusté.
*   **Limites :** Risque important de surapprentissage si le nombre d'arbres est trop grand ou si le taux d'apprentissage n'est pas bridé, et plus difficile à paralléliser.

### **Schéma explicatif**
Illustration de la réduction progressive des résidus d'entraînement au fur et à mesure des étapes de boosting.

    2.  *Marketing :* Prédiction du taux d'engagement ou de clic d'une bannière publicitaire à l'aide de nombreuses variables d'interactions.
*   **Nature des données :** Gère nativement les valeurs manquantes et s'adapte à des structures de données complexes non-linéaires.
*   **Cas peu approprié :** En présence d'un volume extrême de bruit sur la cible (l'algorithme risque d'apprendre spécifiquement les anomalies).
*   **Métriques adaptées :** MAE, RMSE, $R^2$.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La **Régression Logistique** (*Logistic Regression*) est un modèle de l'**apprentissage supervisé (classification)**.

*   **Intuition et étapes :** Malgré son nom, c'est un algorithme de classification binaire (ou multinomiale). Il prend des variables explicatives $X$, calcule une combinaison linéaire puis applique la fonction sigmoïde pour projeter la sortie entre $0$ et $1$, correspondant à la probabilité d'appartenance à la classe positive.
*   **Formule mathématique essentielle :**
    $$P(y=1|X) = \sigma(\beta_0 + \sum_{j=1}^{p} \beta_j X_j) = \frac{1}{1 + e^{-(\beta_0 + \sum_{j=1}^{p} \beta_j X_j)}}$$
    *   $\sigma(z)$ : fonction logistique (sigmoïde).
    *   $\beta_j$ : coefficients de régression appris.
*   **Avantages :** Donne des probabilités bien étalonnées et est extrêmement rapide à entraîner.
*   **Limites :** Suppose une frontière de décision linéaire entre les classes et est sensible à la colinéarité des variables.

### **Schéma explicatif**

### **Contextes d'utilisation**

    2.  *Santé :* Évaluation de la probabilité d'avoir une pathologie donnée à partir de facteurs biologiques standardisés.
*   **Nature des données :** Variables explicatives d'échelle similaire (nécessite une standardisation) et absence d'outliers sévères.
*   **Cas peu approprié :** Frontière de décision hautement complexe, non-linéaire ou imbriquée (comme des cercles concentriques).
*   **Métriques adaptées :** Score F1, Précision, Rappel, Courbe ROC-AUC.

### **Présentation**
L'algorithme des **k-Plus Proches Voisins** (*K-Nearest Neighbors - KNN*) est une méthode de l'**apprentissage supervisé (classification/régression)** non-paramétrique basée sur l'instance.

*   **Intuition et étapes :** Il ne construit pas de modèle mathématique explicite lors de la phase d'apprentissage (c'est un algorithme "paresseux"). Pour prédire la classe d'un nouvel échantillon, il cherche les $k$ observations les plus proches dans l'espace des descripteurs, puis attribue la classe majoritaire parmi ces voisins.
*   **Formule mathématique essentielle (Distance Euclidienne par défaut) :**
    $$d(x, x') = \sqrt{\sum_{j=1}^{p} (x_j - x'_j)^2}$$
    *   $x, x'$ : vecteurs de caractéristiques des deux points comparés.
*   **Avantages :** Très intuitif à comprendre, n'impose aucune hypothèse sur la forme géométrique de la frontière.
*   **Limites :** Coût computationnel élevé en phase de test (calcul de toutes les distances) et sensibilité au fléau de la dimensionnalité (*curse of dimensionality*).

### **Schéma explicatif**

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'algorithme **Naïve Bayes** (ou Classifieur Bayésien Naïf) est une méthode d'**apprentissage supervisé (classification)** basée sur le théorème de Bayes avec l'hypothèse forte (dite "naïve") d'indépendance conditionnelle entre chaque paire de caractéristiques sachant la classe.

*   **Intuition et étapes :** Données en entrée ($X$, $y$) $\rightarrow$ Phase d'apprentissage (estimation des probabilités a priori et des probabilités conditionnelles de chaque variable pour chaque classe) $\rightarrow$ Phase de prédiction (application du théorème de Bayes pour estimer la classe maximisant la probabilité a posteriori).
*   **Formule mathématique essentielle :**
    $$P(y = c | X) \propto P(y = c) \prod_{i=1}^{p} P(X_i | y = c)$$
    *   $P(y = c | X)$ : probabilité *a posteriori* d'appartenir à la classe $c$.
    *   $P(y = c)$ : probabilité *a priori* de la classe $c$.
    *   $P(X_i | y = c)$ : vraisemblance conditionnelle de la caractéristique $i$ sachant la classe.
*   **Avantages :** Extrêmement rapide à entraîner, requiert peu de données d'entraînement et se comporte très bien sur de la classification de texte.
*   **Limites :** L'hypothèse d'indépendance absolue des descripteurs est rarement vérifiée en pratique et dégrade la calibration des probabilités prédites.

Visualisation simplifiée de la forme gaussienne des distributions de probabilité conditionnelle pour deux caractéristiques continues d'une classe.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'**Arbre de Décision pour la classification** (*Decision Tree Classifier*) est un algorithme de l'**apprentissage supervisé (classification)**.

*   **Intuition et étapes :** Il fragmente l'espace géométrique des descripteurs en sous-ensembles rectangulaires disjoints par des divisions logiques et successives, puis associe à chaque segment la classe majoritaire des observations qui s'y trouvent.
*   **Formule mathématique essentielle (Mesure d'Impureté de Gini) :**
    $$H(Q) = \sum_{k} p_k (1 - p_k)$$

*   **Avantages :** Très lisible, transparent (règles logiques claires) et ne nécessite pas de mise à l'échelle des données.
*   **Limites :** Tendance naturelle et rapide au surapprentissage et forte instabilité aux variations légères des données d'entraînement.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La **Machine à Vecteurs de Support pour la classification** (*Support Vector Machine - SVM*) est un algorithme d'**apprentissage supervisé (classification)** non-paramétrique.

*   **Intuition et étapes :** Il cherche à trouver l'hyperplan de séparation optimal qui maximise la marge (la distance) entre les classes d'apprentissage. En cas de données non-linéairement séparables, il projette les données dans un espace de dimension supérieure via l'astuce du noyau (*kernel trick*) pour y trouver une séparation linéaire.
*   **Formule mathématique essentielle (Problème de maximisation de marge avec pénalité soft-margin) :**
    $$\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^{n} \xi_i$$
    Sujet à : $y_i (w^T \phi(X_i) + b) \ge 1 - \xi_i$ et $\xi_i \ge 0$
    *   $w, b$ : paramètres de l'hyperplan.
    *   $C$ : hyperparamètre de compromis entre la largeur de la marge et la tolérance aux erreurs d'entraînement.
    *   $\xi_i$ : variables d'écart (*slack variables*).
    *   $\phi(\cdot)$ : fonction de transformation noyau.
*   **Avantages :** Très performant sur les espaces à haute dimension et efficace lorsque le nombre de caractéristiques dépasse le nombre d'observations.
*   **Limites :** Choix complexe et sensible de l'hyperparamètre $C$ et du type de noyau, et apprentissage coûteux sur des jeux de données de très grande taille.

### **Schéma explicatif**

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
Le **Gradient Boosting pour la classification** (*Gradient Boosting Classifier*) est une technique d'ensemble de la famille de l'**apprentissage supervisé (classification)**.

*   **Intuition et étapes :** Similaire à la régression, il construit des arbres séquentiellement en cherchant à minimiser la log-loss globale. Chaque arbre subséquent s'ajuste sur le gradient de la fonction de perte par rapport aux prédictions de probabilités courantes.
*   **Formule mathématique essentielle (Log-Loss ou Entropie Croisée binaire) :**
    $$L(y, p) = -\frac{1}{n} \sum_{i=1}^{n} [y_i \log(p_i) + (1 - y_i) \log(1 - p_i)]$$
    L'optimisation utilise le gradient de $L(y, p)$ pour guider la structure des nouveaux arbres faibles.
*   **Avantages :** Extrêmement précis, gère naturellement les valeurs manquantes et s'adapte sans effort aux relations complexes.
*   **Limites :** Temps d'entraînement séquentiel non parallélisable naturellement et forte sensibilité au bruit (surapprentissage).

### **Présentation**
**AdaBoost** (*Adaptive Boosting*) est un algorithme d'ensemble de la famille de l'**apprentissage supervisé (classification)**.

*   **Intuition et étapes :** Il combine séquentiellement des classifieurs faibles (généralement des arbres de décision de profondeur 1, ou *decision stumps*). À chaque itération, les poids des exemples mal classés par l'arbre précédent sont augmentés, forçant l'arbre suivant à se concentrer sur les cas difficiles. La décision finale est une somme pondérée des prédictions de chaque classifieur faible.
*   **Formule mathématique essentielle (Mise à jour des poids d'entraînement) :**

    *   $\alpha_t$ : poids attribué au classifieur faible $h_t$ (basé sur son taux d'erreur).

*   **Avantages :** Très simple à mettre en œuvre, polyvalent et peu sujet au surapprentissage si le niveau de bruit reste faible.
*   **Limites :** Très sensible aux valeurs aberrantes (*outliers*) et aux bruits sur les étiquettes car il augmente de manière exponentielle les poids associés à ces erreurs.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
**K-Means** (ou algorithme des K-Moyennes) est une méthode d'**apprentissage non supervisé (partitionnement / clustering)**.

*   **Intuition et étapes :** Il regroupe les données en $K$ clusters homogènes. Après une initialisation des $K$ centroïdes, il alterne deux étapes jusqu'à convergence : 1) assigner chaque observation au centroïde le plus proche ; 2) recalculer la position de chaque centroïde comme le barycentre (la moyenne) des observations de son groupe.
*   **Formule mathématique essentielle (Inertie Intra-classe à minimiser) :**
    $$J = \sum_{k=1}^{K} \sum_{x_i \in C_k} \|x_i - \mu_k\|^2$$
    *   $C_k$ : ensemble des points du cluster $k$.
    *   $\mu_k$ : centroïde (vecteur moyen) du cluster $k$.
*   **Avantages :** Très simple à comprendre, facile à implémenter et linéaire en temps de calcul vis-à-vis du volume de données.
*   **Limites :** L'utilisateur doit spécifier à l'avance la valeur du paramètre $K$, et l'algorithme est sensible à l'initialisation initiale ainsi qu'aux formes non sphériques.

### **Schéma explicatif**
Visualisation de la convergence et des centroïdes finaux trouvés sur nos données de regroupement.

### **Présentation**
La **Classification Ascendante Hiérarchique (CAH)** (*Agglomerative Clustering*) est un algorithme de la famille de l'**apprentissage non supervisé (clustering)**.

*   **Intuition et étapes :** C'est une stratégie ascendante (du bas vers le haut) : elle part d'une situation où chaque observation forme son propre cluster isolé, puis fusionne itérativement les paires de clusters les plus proches selon un critère de liaison (ex. Ward, liaison complète) jusqu'à ce que tous les éléments soient réunis en un groupe unique. L'arbre de fusion résultant est appelé un dendrogramme.
*   **Formule mathématique essentielle (Critère de liaison de Ward) :**
    $$\Delta(A, B) = \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2$$
    *   $n_A, n_B$ : nombres d'observations dans les clusters $A$ et $B$.
    *   $\mu_A, \mu_B$ : centroïdes des clusters $A$ et $B$.
*   **Avantages :** Aucun besoin de spécifier à priori le nombre de clusters (on peut choisir la découpe finale sur le dendrogramme) et produit une taxonomie hiérarchique exploitable.
*   **Limites :** Coût algorithmique en $O(n^3)$ ce qui le rend inutilisable sur des jeux de données volumineux.

### **Schéma explicatif**
Visualisation d'un dendrogramme montrant le processus d'agrégation hiérarchique.

```python
from scipy.cluster.hierarchy import dendrogram, linkage

# Calcul de la matrice de liaison hiérarchique
Z_link = linkage(X_clu[:40], method='ward')

plt.figure(figsize=(8, 4))
dendrogram(Z_link)
plt.title('Dendrogramme partiel de la classification hiérarchique (40 points)')
plt.xlabel('Index des points')
plt.ylabel('Distance de Ward (Coût de fusion)')
plt.grid(True, linestyle='--')
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Génétique :* Regroupement taxonomique de gènes ou de micro-organismes ayant des expressions similaires.
    2.  *Marketing :* Identification de typologies de consommateurs basées sur des profils socio-économiques complexes pour du ciblage ciblé.
*   **Nature des données :** Données continues ou catégorielles associées à une distance de similarité définie.
*   **Cas peu approprié :** Ensembles de données massifs de plusieurs centaines de milliers de lignes.
*   **Métriques adaptées :** Coefficient Cophérétique, Coefficient de Silhouette.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'algorithme **DBSCAN** (*Density-Based Spatial Clustering of Applications with Noise*) est une méthode de la famille de l'**apprentissage non supervisé (clustering)** basée sur la densité locale des points.

*   **Intuition et étapes :** DBSCAN définit les groupes comme des zones denses séparées par des zones de faible densité. Pour chaque point, il compte le nombre de voisins dans un rayon donné (Epsilon). Si ce nombre dépasse un seuil minimum (MinPts), le point est qualifié de point central (*Core Point*) et amorce un cluster. Les points proches des Core Points mais n'ayant pas assez de voisins sont des points de bordure (*Border Points*). Les autres sont rejetés en tant que bruit (*Noise Points*).
*   **Formule mathématique essentielle (Définition du voisinage) :**
    $$N_{\epsilon}(p) = \{q \in D \mid \text{dist}(p, q) \le \epsilon\}$$
    Un point $p$ est un point central si et seulement si $|N_{\epsilon}(p)| \ge \text{MinPts}$.
*   **Avantages :** Découvre des groupes de formes géométriques arbitraires et gère explicitement le bruit/les valeurs aberrantes.
*   **Limites :** Sensible aux écarts de densité au sein du jeu de données, et le choix des hyperparamètres $\epsilon$ et $\text{MinPts}$ est délicat.

### **Schéma explicatif**
Visualisation de la notion de densité DBSCAN (distinction entre points centraux, points de bordure, et bruit).

*   **Nature des données :** Données géographiques ou numériques continues. La mesure de distance doit être standardisée.
*   **Cas peu approprié :** Données ayant des densités d'observations très hétérogènes (où un même $\epsilon$ ne peut pas convenir pour tous les groupes).
*   **Métriques adaptées :** Coefficient de Silhouette (uniquement sur les points affectés), ratio de bruit identifié.

### **Présentation**
Le modèle de **Mélange de Gaussiennes** (*Gaussian Mixture Models - GMM*) est un algorithme probabiliste de l'**apprentissage non supervisé (clustering)**.

*   **Intuition et étapes :** GMM suppose que toutes les données proviennent d'un mélange d'un nombre fini de distributions normales de paramètres inconnus. Contrairement à K-Means qui réalise des affectations strictes (*hard clustering*), le modèle GMM calcule pour chaque échantillon la probabilité d'appartenance à chaque composante gaussienne (*soft clustering*). Il utilise l'algorithme d'Espérance-Maximisation (EM) pour apprendre les moyennes, matrices de covariance et poids de chaque composante.
*   **Formule mathématique essentielle (Densité du mélange) :**

    *   $\mathcal{N}(x \mid \mu_k, \Sigma_k)$ : densité d'une loi normale multidimensionnelle.
*   **Avantages :** Permet des clusters de formes ellipsoïdales et fournit une mesure probabiliste d'appartenance.
*   **Limites :** Sensible aux extrema locaux de vraisemblance et instable si le nombre d'échantillons est trop faible par rapport à la dimension.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'**Analyse en Composantes Principales (ACP)** (*Principal Component Analysis - PCA*) est un algorithme de l'**apprentissage non supervisé (réduction de dimension)**.

*   **Intuition et étapes :** L'objectif est de projeter un ensemble de variables corrélées sur un sous-espace de dimension inférieure constitué de variables non corrélées appelées "composantes principales". Ces composantes sont orientées selon les directions de variance maximale des données. L'algorithme calcule la matrice de covariance, ses valeurs propres et vecteurs propres correspondants.
*   **Formule mathématique essentielle (Calcul de la variance maximale) :**
    $$w_1 = \arg\max_{\|w\|=1} \operatorname{Var}(w^T X) = \arg\max_{\|w\|=1} (w^T \Sigma w)$$
    *   $\Sigma$ : matrice de covariance des variables d'entrée.
    *   $w_1$ : premier vecteur propre (direction du premier axe).
*   **Avantages :** Élimine la multicolinéarité et réduit drastiquement la dimensionnalité tout en conservant le maximum d'information (variance).
*   **Limites :** Les nouvelles variables projetées perdent leur interprétabilité physique directe et l'analyse est sensible aux échelles initiales des données.

### **Schéma explicatif**
Visualisation de la variance capturée le long des axes de projection principaux.

```python
from sklearn.decomposition import PCA

pca_visu = PCA(n_components=2).fit(X_vin)

plt.figure(figsize=(7, 4.5))
plt.bar(['PC1', 'PC2'], pca_visu.explained_variance_ratio_, color='teal', alpha=0.8)
plt.ylabel('Ratio de Variance Expliquée')
plt.title('Variance capturée par les deux premières composantes principales')
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Finance :* Compression du spectre de taux d'intérêt de différentes maturités en 3 composantes principales (niveau, pente, courbure).
    2.  *Gestion des données :* Étape préparatoire de réduction de dimension avant l'apprentissage de modèles supervisés complexes afin de réduire le temps de calcul et d'éviter le fléau de la dimensionnalité.
*   **Nature des données :** Variables d'entrée continues. Nécessite obligatoirement un centrage et une réduction d'échelle (Standardisation).
*   **Cas peu approprié :** Données caractérisées par des structures non-linéaires imbriquées complexes (ex. variété en forme de spirale).
*   **Métriques adaptées :** Taux de variance expliquée cumulée.

### **Analyse de l'expérience Python**
L'expérience normalise le jeu de données `load_wine` puis projette les variables explicatives sur ses 2 axes principaux.

### **Présentation**
L'algorithme **t-SNE** (*t-Distributed Stochastic Neighbor Embedding*) est une méthode de la famille de l'**apprentissage non supervisé (réduction de dimension)** principalement conçue pour la visualisation de données de grande dimension.

*   **Intuition et étapes :** t-SNE convertit les distances euclidiennes entre les points dans l'espace d'origine en probabilités conditionnelles représentant des similitudes. Il projette ensuite les points dans un espace de dimension inférieure (2D ou 3D) en minimisant la divergence de Kullback-Leibler entre les distributions de probabilités de l'espace d'origine et de l'espace cible, en utilisant une distribution t de Student dans l'espace cible pour éviter le problème d'entassement.
*   **Formule mathématique essentielle (Divergence KL à minimiser) :**
    $$KL(P || Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$
    *   $p_{j|i}$ : probabilité de similarité entre les points $i$ et $j$ dans l'espace d'origine.
    *   $q_{j|i}$ : probabilité de similarité entre les points $i$ et $j$ dans l'espace projeté.
*   **Avantages :** Remarquable pour préserver les structures locales et révéler des clusters imbriqués complexes.
*   **Limites :** Processus stochastique non déterministe, ne préserve pas les distances globales et temps de calcul très élevé.

### **Présentation**
L'algorithme **UMAP** (*Uniform Manifold Approximation and Projection*) est une méthode d'**apprentissage non supervisé (réduction de dimension)** fondée sur la géométrie riemannienne et la topologie algébrique.

*   **Intuition et étapes :** UMAP suppose que les données résident sur une variété topologique locale. Il construit une représentation floue d'un complexe simplicial basé sur les distances locales des points, puis optimise une projection en basse dimension en minimisant l'entropie croisée floue entre les représentations topologiques de l'espace d'origine et de l'espace cible.
*   **Formule mathématique essentielle (Entropie croisée floue) :**
    $$d_{\text{fuzzy}}(E, P) = \sum_{e \in E} \left[ \mu(e) \log \frac{\mu(e)}{\nu(e)} + (1 - \mu(e)) \log \frac{1 - \mu(e)}{1 - \nu(e)} \right]$$
    *   $\mu(e)$ : appartenance d'une arête dans l'espace d'origine.
    *   $\nu(e)$ : appartenance d'une arête dans l'espace projeté.
*   **Avantages :** Conserve beaucoup mieux la structure globale des données que t-SNE, tout en étant considérablement plus rapide et scalable.
*   **Limites :** Théorie mathématique très abstraite et complexe, et sensibilité aux hyperparamètres tels que le nombre de voisins (`n_neighbors`).

### **Schéma explicatif**

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'algorithme **Isolation Forest** (Forêt d'Isolation) est une méthode d'**apprentissage non supervisé (détection d'anomalies)**.

*   **Intuition et étapes :** Au lieu de modéliser les profils normaux des données d'entraînement, il isole explicitement les anomalies. Il construit une forêt d'arbres binaires de décision de manière aléatoire. Comme les anomalies possèdent des valeurs d'attributs inhabituelles, elles nécessitent beaucoup moins de découpes aléatoires pour être isolées dans l'arbre, ce qui se traduit par des chemins d'accès significativement plus courts dans la structure de l'arbre.
*   **Formule mathématique essentielle (Score d'anomalie d'une observation) :**
    $$s(x, n) = 2^{-\frac{E(h(x))}{c(n)}}$$
    *   $h(x)$ : longueur du chemin pour isoler l'observation $x$.
    *   $E(h(x))$ : moyenne des longueurs de chemin sur la forêt d'arbres.
    *   $c(n)$ : longueur moyenne théorique d'un chemin d'isolement pour $n$ points.
    *   Un score $s$ proche de 1 indique une forte probabilité d'anomalie.
*   **Avantages :** Efficacité computationnelle remarquable (complexité linéaire) et très robuste au phénomène de masquage (*masking effect*).
*   **Limites :** Définir de manière précise l'hyperparamètre de contamination est indispensable pour calibrer le seuil de décision.

### **Schéma explicatif**
Visualisation de la détection des points marginaux (anomalies) écartés du nuage principal par la Forêt d'Isolation.

    1.  *Finance / Monétique :* Détection de fraudes aux transactions de paiement (comportement d'achat extrêmement éloigné de l'historique du porteur de la carte).
    2.  *Sécurité Réseau :* Détection d'intrusions sur des infrastructures d'hébergement cloud d'après les débits de paquets.
*   **Nature des données :** Données quantitatives multidimensionnelles.
*   **Cas peu approprié :** Détection d'anomalies sur des données temporelles à forte tendance de croissance ou saisonnalité sans prétraitement préalable de stationnarisation.
*   **Métriques adaptées :** Taux de faux positifs, Rappel (détection correcte d'anomalies réelles), aire sous la courbe Precision-Recall (PR-AUC).

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
Le problème des **Bandits Manchots** (*Multi-Armed Bandits*) est un cadre fondamental de l'**apprentissage par renforcement**.

*   **Intuition et étapes :** Un agent se trouve face à plusieurs leviers (bras) de machines à sous. Chaque bras fournit une récompense aléatoire tirée d'une distribution de probabilité inconnue de l'agent. L'objectif est de maximiser la récompense cumulée au fil du temps en apprenant de manière empirique la valeur attendue de chaque bras.
*   **Formule mathématique essentielle (Estimation de la valeur d'une action) :**
    $$Q_t(a) = \frac{\sum_{i=1}^{t-1} R_i \cdot I(A_i = a)}{\sum_{i=1}^{t-1} I(A_i = a)}$$
    *   $Q_t(a)$ : valeur estimée du bras $a$ au temps $t$.
    *   $R_i$ : récompense reçue lors de l'essai $i$.
    *   $I(\cdot)$ : fonction indicatrice renvoyant 1 si l'action choisie était $a$.
*   **Avantages :** Modèle d'optimisation robuste et très simple à implémenter.
*   **Limites :** Ne prend pas en compte le concept d'état d'environnement (pas d'étapes de transition), et l'exploration naïve peut coûter cher.

### **Schéma explicatif**
Visualisation de la distribution des gains réels de chaque bras face aux estimations calculées par l'agent au cours des essais.

```python
plt.figure(figsize=(7, 4))
plt.bar(['Bras 1', 'Bras 2', 'Bras 3'], vraies_probas, color='gray', alpha=0.5, label='Probabilités réelles')
plt.plot(['Bras 1', 'Bras 2', 'Bras 3'], somme/comptes, 'ro-', label='Estimations de l\'agent')
plt.ylabel('Probabilité de récompense')
plt.title('Comparaison entre probabilités réelles et estimations (Multi-Armed Bandits)')
plt.legend()
plt.grid(True, linestyle='--')
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Marketing / Web :* Tests A/B dynamiques pour optimiser la présentation d'une page d'accueil ou de bannières publicitaires selon le taux de clics.
    2.  *Essais Cliniques :* Répartition adaptative de l'attribution de traitements expérimentaux aux patients pour privilégier le traitement se révélant le plus efficace au fur et à mesure.
*   **Nature des données :** Récompenses binaires (clic/pas de clic) ou continues après chaque action.
*   **Cas peu approprié :** Problématiques complexes où les actions en cours affectent directement l'état futur du système (ex. navigation de robot).
*   **Métriques adaptées :** Regret cumulé, récompense moyenne par itération.

### **Analyse de l'expérience Python**
L'expérience simule 500 étapes de choix séquentiels cycliques (sans stratégie d'apprentissage intelligente).
*   **Entrées :** Probabilités réelles de gains définies par `[0.2, 0.5, 0.7]`.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
La stratégie **$\epsilon$-greedy** (Epsilon-Gourmande) est un algorithme classique d'équilibrage de l'exploration et de l'exploitation pour l'**apprentissage par renforcement**.

*   **Intuition et étapes :** À chaque étape, avec une probabilité $\epsilon$, l'agent choisit une action de manière totalement aléatoire pour acquérir de nouvelles connaissances (exploration). Avec une probabilité $1-\epsilon$, il sélectionne la meilleure action connue à ce jour (exploitation).
*   **Formule mathématique essentielle (Règles de décision) :**
    $$A_t = \begin{cases} \text{action aléatoire} & \text{avec probabilité } \epsilon \\ \operatorname{arg\,max}_{a} Q_t(a) & \text{avec probabilité } 1 - \epsilon \end{cases}$$
    *   $\epsilon$ : hyperparamètre d'exploration ($0 \le \epsilon \le 1$).
*   **Avantages :** Extrêmement simple à implémenter et garantit qu'aucune action n'est définitivement oubliée à l'infini.
*   **Limites :** L'exploration est purement uniforme et n'oriente pas l'exploration vers les bras les plus prometteurs.

### **Schéma explicatif**
Visualisation de la répartition entre actions d'exploration et d'exploitation au fil des étapes.

```python
plt.figure(figsize=(7, 4))
plt.pie(comptes, labels=[f'Bras 1 (Choix: {comptes[0]})', f'Bras 2 (Choix: {comptes[1]})', f'Bras 3 (Choix: {comptes[2]})'], autopct='%1.1f%%', colors=['#ff9999','#66b3ff','#99ff99'])
plt.title('Répartition des choix de bras sous stratégie Epsilon-Greedy')
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *E-commerce :* Algorithme de recommandation d'articles recommandant la majorité du temps des succès commerciaux (exploitation) tout en proposant parfois des nouveautés (exploration).
    2.  *Finance :* Sélection d'actifs pour de l'arbitrage de portefeuille à haute fréquence.
*   **Nature des données :** Flux d'interactions en direct avec un environnement.
*   **Cas peu approprié :** Scénarios où l'exploration au hasard a des conséquences réelles catastrophiques (ex. conduite autonome).
*   **Métriques adaptées :** Récompense cumulée, regret cumulé.

### **Analyse de l'expérience Python**
L'expérience simule 1000 tirages avec un $\epsilon=0.1$.
*   **Entrées :** Vraies probabilités `[0.2, 0.5, 0.7]`.
*   **Hyperparamètres :** `epsilon=0.1`.
*   **Sortie :** Nombre de choix : `[25, 49, 926]`.
*   **Conclusion :** Grâce à l'exploitation intelligente, le bras optimal (Bras 3) a été sélectionné l'immense majorité du temps (92.6%), montrant l'efficacité de la stratégie face à un choix purement cyclique.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
Le **Q-learning** est un algorithme d'**apprentissage par renforcement** hors-politique (*off-policy*).

*   **Intuition et étapes :** Il cherche à apprendre de manière itérative la valeur (notée $Q$) de chaque paire (état, action) sans modèle de l'environnement (*model-free*). Il met à jour ses estimations en supposant que l'agent prendra toujours l'action optimale à l'étape suivante, indépendamment de la politique de décision effectivement suivie.
*   **Formule mathématique essentielle (Réquation de mise à jour temporelle de Bellman) :**
    $$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
    *   $s, a$ : état et action courants.
    *   $R, s'$ : récompense immédiate reçue et état suivant atteint.
    *   $\alpha$ : taux d'apprentissage (hyperparamètre).
    *   $\gamma$ : facteur d'actualisation temporelle (hyperparamètre).
*   **Avantages :** Apprend la politique optimale de manière autonome même avec des données générées par une politique exploratoire sous-optimale.
*   **Limites :** Souffre du fléau de la dimensionnalité (la table Q devient trop grande si le nombre d'états explose).

### **Schéma explicatif**
Visualisation de la convergence de la table Q après l'entraînement de l'agent sur l'environnement FrozenLake.

```python
plt.figure(figsize=(8, 4))
plt.imshow(Q, cmap='hot', interpolation='nearest')
plt.colorbar(label='Valeur Q estimée')
plt.title('Carte thermique de la Table Q finale (FrozenLake)')
plt.xlabel('Actions (Gauche, Bas, Droite, Haut)')
plt.ylabel('États (0 à 15)')
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Robotique :* Apprentissage du déplacement d'un bras mécanique articulé dans un labyrinthe simple à états discrets.
    2.  *Logistique :* Optimisation de la gestion dynamique de stocks d'entrepôt en fonction de la demande saisonnière modélisée comme un processus de décision markovien (MDP).
*   **Nature des données :** États, actions, transitions, récompenses discrets.
*   **Cas peu approprié :** Environnements où les variables d'état sont d'ordre continu (ex. maintenir l'équilibre d'un pendule).
*   **Métriques adaptées :** Récompense par épisode, taux de succès pour atteindre la cible.

### **Analyse de l'expérience Python**
L'expérience entraîne un agent par Q-learning sur FrozenLake-v1 (carte déterministe) pour 4000 épisodes.
*   **Entrées :** Environnement FrozenLake-v1.
*   **Hyperparamètres :** `learning_rate=0.3`, `gamma=0.95`.
*   **Sortie :** Politique optimale finale : `[1, 2, 1, 0, 1, 0, 1, 0, 2, 1, 1, 0, 0, 2, 2, 0]`.
*   **Conclusion :** L'agent a appris avec succès les meilleures décisions pour contourner les trous et atteindre l'objectif de manière systématique.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'algorithme **SARSA** (*State-Action-Reward-State-Action*) est une méthode d'**apprentissage par renforcement** sur-politique (*on-policy*).

*   **Intuition et étapes :** Contrairement au Q-learning, SARSA évalue la valeur des paires (état, action) en se basant sur l'action réellement choisie à l'état suivant d'après sa politique d'exploration (par exemple $\epsilon$-greedy), plutôt que de supposer une action future idéale. La mise à jour dépend du quintuplet $(s_t, a_t, r_{t+1}, s_{t+1}, a_{t+1})$.
*   **Formule mathématique essentielle :**
    $$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R + \gamma Q(s', a') - Q(s, a) \right]$$
    *   $a'$ : action effectivement sélectionnée à l'état suivant $s'$ sous la politique courante.
*   **Avantages :** Politique d'apprentissage plus prudente, limitant l'exposition aux risques mortels de l'environnement pendant l'apprentissage.
*   **Limites :** Peut converger vers une politique globale sous-optimale si l'exploration reste très active.

### **Schéma explicatif**
Comparaison des valeurs de Q calculées par SARSA vis-à-vis des trajectoires d'exploration d'un agent.

```python
plt.figure(figsize=(8, 4))
plt.plot(np.max(Q, axis=1), 'bo-', label='Q-Learning (Optimal theorique)')
plt.title('Comparaison de la valeur de l\'état maximum par case')
plt.xlabel('Index de l\'État')
plt.ylabel('Valeur max de Q')
plt.grid(True, linestyle='--')
plt.show()
```

### **Contextes d'utilisation**
*   **Deux cas d'usage concrets :**
    1.  *Pilotage de Drones :* Guidage autonome où les trajectoires dangereuses (collisions) doivent être fortement pénalisées et évitées dès la phase d'apprentissage.
    2.  *Gestion Industrielle :* Chauffage ou refroidissement de serveurs de données où les pointes de température dégradent l'équipement.
*   **Nature des données :** Interactions par épisodes (états, actions discrets).
*   **Cas peu approprié :** Problèmes théoriques purs sans risque physique où l'on veut la trajectoire mathématique optimale absolue sans se soucier des échecs d'entraînement.
*   **Métriques adaptées :** Récompense cumulée, taux de réussite à l'objectif.

### **Analyse de l'expérience Python**
L'expérience applique l'algorithme `SARSA` sur FrozenLake-v1 pour 4000 épisodes.
*   **Entrées :** Environnement FrozenLake-v1.
*   **Hyperparamètres :** `learning_rate=0.3`, `gamma=0.95`, $\epsilon$-décroissant.
*   **Sortie :** Politique SARSA obtenue : `[1, 0, 3, 0, 1, 0, 1, 0, 2, 1, 1, 0, 0, 2, 2, 0]`.
*   **Conclusion :** La politique finale est légèrement différente de celle du Q-learning, car SARSA intègre la prudence liée aux probabilités d'exploration dans ses estimations de valeurs.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### **Présentation**
L'algorithme **DQN** (*Deep Q-Network*) fusionne l'**apprentissage par renforcement** avec l'apprentissage profond (*Deep Learning*).

*   **Intuition et étapes :** Pour surmonter l'impossibilité de maintenir une table Q lorsque l'espace des états est continu ou gigantesque, DQN remplace cette table par un réseau de neurones artificiels qui estime les valeurs $Q(s, a)$ pour n'importe quelle entrée continue $s$. Deux mécanismes clés assurent sa stabilité : l'utilisation d'une mémoire de rejeu d'expérience (*Experience Replay*) pour rompre la corrélation temporelle des échantillons et l'utilisation d'un réseau cible (*Target Network*) mis à jour périodiquement pour éviter la divergence des cibles de calcul.
*   **Formule mathématique essentielle (Calcul de la perte d'entraînement d'un DQN) :**
    $$L(\theta) = \mathbb{E} \left[ \left( R + \gamma \max_{a'} Q(s', a'; \theta^{-}) - Q(s, a; \theta) \right)^2 \right]$$
    *   $\theta$ : poids du réseau de neurones en ligne.
    *   $\theta^{-}$ : poids du réseau cible (figé temporairement).
*   **Avantages :** Capacité unique à résoudre des environnements visuels et continus complexes (ex. jeux Atari, robotique avancée).
*   **Limites :** Très difficile à stabiliser (hyperparamètres de réseaux complexes) et requiert un temps d'entraînement considérable.

### **Schéma explicatif**
Illustration conceptuelle du calcul de l'erreur quadratique de l'équation de Bellman (perte temporelle).

*   **Deux cas d'usage concrets :**
    1.  *Automobile :* Algorithmes de décision pour la conduite autonome (vitesse de croisière, changements de file d'après les capteurs continus).
    2.  *Gestion d'Infrastructures :* Régulation en temps réel de la climatisation et de la consommation d'énergie de datacenters géants.
*   **Nature des données :** Espaces d'états multidimensionnels continus (flux vidéo, signaux de capteurs physiques).
*   **Cas peu approprié :** Problèmes de décision simples et discrets où une table classique ou une approche tabulaire classique (Q-learning) est largement suffisante et moins coûteuse en ressources de calcul.
*   **Métriques adaptées :** Récompense cumulée par épisode, valeur de perte du réseau de neurones, stabilité des valeurs Q.
